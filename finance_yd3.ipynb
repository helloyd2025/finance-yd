# 데이터 스케일링

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# 범주형 및 수치형 컬럼 구분
categorical_features = ['업종(중분류)', '주소지시군구', '외감구분']
numerical_features = [col for col in df_feats.columns if col not in categorical_features]

# 전처리 파이프라인 정의
# 수치형 컬럼: StandardScaler 적용
# 범주형 컬럼: OneHotEncoder 적용 (unknown_value='ignore'를 사용하여 학습 데이터에 없는 범주 처리)
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# 전처리 적용
df_processed_feats = preprocessor.fit_transform(df_feats)

# 처리된 피처의 형태 확인 (희소 행렬일 수 있음)
print("Shape of processed features:", df_processed_feats.shape)
```

## 학습 데이터 불균형 해소

학습 데이터 자체가 합성 데이터이므로 불균형 해소를 위해 SMOTE를 적용하는 것이 적절한지는 의문.

1만 개 미만인 클래스만 본 개수의 3-9배로 합성.

```python
print("Class distribution for '기업신용평가등급(구간화)':\n", y_train_grade.value_counts().sort_index())
print("\nClass distribution for '모형개발용Performance(향후1년내부도여부)':\n", y_train_performance.value_counts().sort_index())
```

```python
from imblearn.over_sampling import SMOTE

def sampling_func(y):
    unique, counts = np.unique(y, return_counts=True)
    strategy = {}
    for cls, cnt in zip(unique, counts):
        if cnt < 10000:
            target = cnt * 3
            if target <= 10000:
                target = cnt * 9
            strategy[cls] = int(target)
    return strategy

smote = SMOTE(random_state=RANDOM_SEED, k_neighbors=4, sampling_strategy=sampling_func, n_jobs=-1)
X_train_grade_smote, y_train_grade_smote = smote.fit_resample(X_train_grade, y_train_grade)
X_train_performance_smote, y_train_performance_smote = smote.fit_resample(X_train_performance, y_train_performance)

print("Class distribution for '기업신용평가등급(구간화)':\n", y_train_grade_smote.value_counts().sort_index())
print("Class distribution for '모형개발용Performance(향후1년내부도여부)' after SMOTE:\n", y_train_performance_smote.value_counts().sort_index())
```

```python
from lightgbm import LGBMClassifier
from sklearn.model_selection import RandomizedSearchCV

# 멀티 클래스
lgbm_grade = LGBMClassifier(objective='multiclass',
                            num_class=len(y_train_grade_smote.unique()),
                            metric='multi_logloss',
                            class_weight='balanced'
                            n_jobs=-1,
                            random_state=RANDOM_SEED,
                            device='gpu')

param_dist_lgbm = {
    'n_estimators': stats.randint(100, 500),
    'learning_rate': stats.loguniform(0.01, 0.1),
    'lambda_l1': stats.uniform(0.0, 0.2),
    'lambda_l2': stats.uniform(0.0, 0.2),
    'boosting_type': ['dart', 'goss'],
    'num_leaves': stats.randint(20, 60),
    'max_depth': stats.randint(5, 15),
    'min_child_samples': stats.randint(20, 100),
    'subsample': stats.uniform(0.6, 0.4),
    'colsample_bytree': stats.uniform(0.6, 0.4)
}

random_search_lgbm_grade = RandomizedSearchCV(estimator=lgbm_grade,
                                              param_distributions=param_dist_lgbm,
                                              n_iter=50, cv=3,
                                              scoring='f1_macro',
                                              random_state=RANDOM_SEED,
                                              verbose=1, n_jobs=-1)

print("\nStarting RandomizedSearchCV for '기업신용평가등급(구간화)'...")
random_search_lgbm_grade.fit(X_train_grade_smote, y_train_grade_smote)
```

```python
print("Finished RandomizedSearchCV for '기업신용평가등급(구간화)'.")
print("Best parameters for grade model:", random_search_lgbm_grade.best_params_)
print("Best cross-validation accuracy for grade model:", random_search_lgbm_grade.best_score_)

best_grade_model = random_search_lgbm_grade.best_estimator_
y_pred_grade = best_grade_model.predict(X_test_grade)
print("\nEvaluation for '기업신용평가등급(구간화)' model:")
print("Accuracy:", accuracy_score(y_test_grade, y_pred_grade))
print("Classification Report:\n", classification_report(y_test_grade, y_pred_grade))
```
