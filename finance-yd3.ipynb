| 모델                        | 특징              | MICE에서의 효과                                                                  |

```python
target_cols = ['기업신용평가등급(구간화)', '모형개발용Performance(향후1년내부도여부)']

df_targets = df_dropped_na[target_cols]
df_feats = df_dropped_na[[c for c in df_dropped_na.columns if c not in target_cols]]
```

# 데이터 스케일링

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# 범주형 및 수치형 컬럼 구분
categorical_features = ['업종(중분류)', '주소지시군구', '외감구분']
numerical_features = [col for col in df_feats.columns if col not in categorical_features]

# 전처리 파이프라인 정의
# 수치형 컬럼: StandardScaler 적용
# 범주형 컬럼: OneHotEncoder 적용 (unknown_value='ignore'를 사용하여 학습 데이터에 없는 범주 처리)
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ])

# 전처리 적용
df_processed_feats = preprocessor.fit_transform(df_feats)

# 처리된 피처의 형태 확인 (희소 행렬일 수 있음)
print("Shape of processed features:", df_processed_feats.shape)
```

# 모델 구축

```python
print("Missing values in '업종(중분류)':", df_feats['업종(중분류)'].isna().sum())
print("Missing values in '주소지시군구':", df_feats['주소지시군구'].isna().sum())
```

## 데이터 분할

전처리된 피처 데이터 (`df_processed_feats`)와 타겟 데이터 (`df_targets`)를 학습 세트와 테스트 세트로 분할.


```python
from sklearn.model_selection import train_test_split

y_grade = df_targets['기업신용평가등급(구간화)']
y_performance = df_targets['모형개발용Performance(향후1년내부도여부)']

X_train_grade, X_test_grade, y_train_grade, y_test_grade = train_test_split(
    df_processed_feats, y_grade, test_size=0.2, random_state=42, stratify=y_grade)

X_train_performance, X_test_performance, y_train_performance, y_test_performance = train_test_split(
    df_processed_feats, y_performance, test_size=0.2, random_state=42, stratify=y_performance)

print("Shape of X_train_grade:", X_train_grade.shape)
print("Shape of y_train_grade:", y_train_grade.shape)
print("Shape of X_test_grade:", X_test_grade.shape)
print("Shape of y_test_grade:", y_test_grade.shape)
print("\nShape of X_train_performance:", X_train_performance.shape)
print("Shape of y_train_performance:", y_train_performance.shape)
print("Shape of X_test_performance:", X_test_performance.shape)
print("Shape of y_test_performance:", y_test_performance.shape)
```

## 모델 선택


```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression

# '기업신용평가등급(구간화)' 예측을 위한 다중 클래스 분류 모델 후보: RandomForestClassifier, GradientBoostingClassifier
# '모형개발용Performance(향후1년내부도여부)' 예측을 위한 이진 분류 모델 후보: LogisticRegression, RandomForestClassifier, GradientBoostingClassifier
```

## 모델 학습


```python
model_grade = RandomForestClassifier(n_estimators=100, random_state=42)
model_grade.fit(X_train_grade, y_train_grade)
print("RandomForestClassifier model for '기업신용평가등급(구간화)' trained successfully.")

model_performance = GradientBoostingClassifier(n_estimators=100, random_state=42)
model_performance.fit(X_train_performance, y_train_performance)
print("GradientBoostingClassifier model for '모형개발용Performance(향후1년내부도여부)' trained successfully.")
```

## 모델 평가


```python
from sklearn.metrics import accuracy_score, classification_report, precision_score, recall_score, f1_score, roc_auc_score

# Evaluate model_grade (Multi-class Classification)
y_pred_grade = model_grade.predict(X_test_grade)
print("Evaluation for '기업신용평가등급(구간화)' model:")
print("Accuracy:", accuracy_score(y_test_grade, y_pred_grade))
print("Classification Report:\n", classification_report(y_test_grade, y_pred_grade))

# Evaluate model_performance (Binary Classification)
y_pred_performance = model_performance.predict(X_test_performance)
y_pred_proba_performance = model_performance.predict_proba(X_test_performance)[:, 1]

print("\nEvaluation for '모형개발용Performance(향후1년내부도여부)' model:")
print("Accuracy:", accuracy_score(y_test_performance, y_pred_performance))
print("Precision:", precision_score(y_test_performance, y_pred_performance))
print("Recall:", recall_score(y_test_performance, y_pred_performance))
print("F1-score:", f1_score(y_test_performance, y_pred_performance))
print("ROC-AUC:", roc_auc_score(y_test_performance, y_pred_proba_performance))
```
